```python
import numpy as np
import matplotlib.pyplot as plt
from fixedpoint import FixedPoint 
```

numbers from 0 to 1

```python
x = np.linspace(-1, 1, 1000)
```

```python
x_iq16 = np.zeros(len(x))
ff_iq16 =np.zeros(len(x))

x_iq24 =np.zeros(len(x))
ff_iq24 =np.zeros(len(x))


for i in range(len(x)):
   x_iq16[i] = (FixedPoint(x[i], m=16, n=16))
   x_iq24[i] = (FixedPoint(x[i], m=8, n=24))
   ff_iq16[i] = (float(x_iq16[i]))
   ff_iq24[i] = (float(x_iq24[i]))

plt.subplot(311)
plt.plot(x - ff_iq16, '.')
plt.subplot(312)
plt.plot(x - ff_iq24, '.')
plt.subplot(313)
plt.plot(ff_iq24 - ff_iq16, '.')
plt.show()
```
